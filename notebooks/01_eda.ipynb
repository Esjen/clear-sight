```python
import json
import pandas as pd
import glob

base = "../data/ember_2018/ember_dataset_2018_2/ember2018"
files = sorted(glob.glob(f"{base}/*.jsonl"))

for filepath in files:
    labels = []
    with open(filepath, "r") as f:
        for line in f:
            labels.append(json.loads(line)["label"])
    print(f"\n{filepath.split('/')[-1]} — total: {len(labels)}")
    print(pd.Series(labels).value_counts())
```

```python
import json
import pandas as pd
import glob

base = "../data/ember_2018/ember_dataset_2018_2/ember2018"
train_files = sorted(glob.glob(f"{base}/train_features_*.jsonl"))

data = []
for filepath in train_files:
    with open(filepath, "r") as f:
        for line in f:
            row = json.loads(line)
            if row["label"] != -1:   # drop unlabeled rows
                data.append(row)

df_train = pd.DataFrame(data)
print(f"Combined training set: {df_train.shape}")
print(df_train['label'].value_counts())
```
